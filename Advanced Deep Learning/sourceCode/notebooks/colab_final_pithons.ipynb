## Blend and Submission

```python
from pathlib import Path
import csv, sys, zipfile
import numpy as np
from tqdm import tqdm
from adl_anomaly.q8rle import q8rle_to_float_matrix, float_matrix_to_q8rle

def raise_csv_field_limit():
    limit = sys.maxsize
    while True:
        try:
            csv.field_size_limit(limit)
            return
        except OverflowError:
            limit //= 10

def read_submission(path):
    raise_csv_field_limit()
    path = Path(path)
    if path.suffix == ".zip":
        zf = zipfile.ZipFile(path)
        names = [n for n in zf.namelist() if n.endswith(".csv")]
        if len(names) != 1:
            zf.close()
            raise ValueError(f"{path} must contain exactly one CSV, found {len(names)}")
        handle = zf.open(names[0], "r")
        close_owner = zf
    else:
        handle = path.open("rb")
        close_owner = None
    try:
        reader = csv.DictReader(line.decode("utf-8") for line in handle)
        if reader.fieldnames is None:
            raise ValueError(f"{path} must have a header")
        id_col = "ID" if "ID" in reader.fieldnames else "image_id"
        label_col = "Label" if "Label" in reader.fieldnames else "score"
        rows = {row[id_col]: row[label_col] for row in reader}
    finally:
        handle.close()
        if close_owner is not None:
            close_owner.close()
    return rows

def blend_submissions(inputs, weights, output_csv, make_zip=True):
    weights = np.asarray(weights, dtype=np.float32)
    weights = weights / weights.sum()
    submissions = [read_submission(p) for p in inputs]
    image_ids = list(submissions[0].keys())
    expected = set(image_ids)
    for p, rows in zip(inputs[1:], submissions[1:]):
        if set(rows.keys()) != expected:
            raise ValueError(f"{p} has different image IDs")

    output_csv = Path(output_csv)
    output_csv.parent.mkdir(parents=True, exist_ok=True)
    with output_csv.open("w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["ID", "Label"])
        for image_id in tqdm(image_ids, desc="Blend submissions"):
            blended = None
            for weight, rows in zip(weights, submissions):
                decoded = q8rle_to_float_matrix(rows[image_id])
                blended = decoded * float(weight) if blended is None else blended + decoded * float(weight)
            writer.writerow([image_id, float_matrix_to_q8rle(np.clip(blended, 0.0, 1.0))])

    if make_zip:
        zip_path = output_csv.with_suffix(".zip")
        with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
            zf.write(output_csv, arcname=output_csv.name)
        return zip_path
    return output_csv

out_zip = blend_submissions(
    ["/content/submission_ab_uncertainty.csv", "/content/submission_segformer.csv"],
    [0.55, 0.45],
    "/content/submission_supervised_blend.csv",
    make_zip=True,
)

print("Submission zip:", out_zip)
```
